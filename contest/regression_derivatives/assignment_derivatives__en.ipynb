### Gradient descent for solving a real task
The following function allows you to find optimal parameter values using gradient descent:

```python
def get_w_by_grad(X, Y, w_0, loss_mode="mse", reg_mode=None, lr=0.05, n_steps=100, reg_coeff=0.05):
    if loss_mode == "mse":
        loss_function = LossAndDerivatives.mse
        loss_derivative = LossAndDerivatives.mse_derivative
    elif loss_mode == "mae":
        loss_function = LossAndDerivatives.mae
        loss_derivative = LossAndDerivatives.mae_derivative
    else:
        raise ValueError("Unknown loss function. Available loss functions: `mse`, `mae`")

    if reg_mode is None:
        reg_function = LossAndDerivatives.no_reg
        reg_derivative = LossAndDerivatives.no_reg_derivative  # lambda w: np.zeros_like(w)
    elif reg_mode == "l2":
        reg_function = LossAndDerivatives.l2_reg
        reg_derivative = LossAndDerivatives.l2_reg_derivative
    elif reg_mode == "l1":
        reg_function = LossAndDerivatives.l1_reg
        reg_derivative = LossAndDerivatives.l1_reg_derivative
    else:
        raise ValueError("Unknown regularization mode. Available modes: `l1`, `l2`, None")

    w = w_0.copy()

    for i in range(n_steps):
        empirical_risk = loss_function(X, Y, w) + reg_coeff * reg_function(w)
        gradient = loss_derivative(X, Y, w) + reg_coeff * reg_derivative(w)
        gradient_norm = np.linalg.norm(gradient)
        if gradient_norm > 5.0:
            gradient = gradient / gradient_norm * 5.0
        w -= lr * gradient

        if i % 25 == 0:
            print(f"Step={i}, loss={empirical_risk},\ngradient values={gradient}\n")
    return w
```

### Comparing the ways to solve it

Those four derivatives are enough to solve the regression problem in several ways, and the ways
differ both in what they cost and in what they converge to. This section compares them on the
[UCI Wine Quality](https://archive.ics.uci.edu/dataset/186/wine+quality) data set: the red and the
white subset pooled, eleven physicochemical measurements plus the wine's colour as a twelfth
feature, and the quality score as the target.

The reference point is the matrix formula, which solves least squares in closed form:
$$
w^* = (X^\top X)^{-1} X^\top y,
\qquad
w^*_\lambda = (X^\top X + \lambda n I)^{-1} X^\top y,
$$
the second being the minimiser of $\frac{1}{n}\|Xw - y\|^2 + \lambda\|w\|^2$ — the ridge solution
the L2 run should converge to.

```python
"""
If you are using Google Colab, uncomment the next two lines to download the wine data
"""
# !wget https://raw.githubusercontent.com/girafe-ai/ml-course/refs/heads/homeworks/contest/regression_derivatives/winequality-red.csv
# !wget https://raw.githubusercontent.com/girafe-ai/ml-course/refs/heads/homeworks/contest/regression_derivatives/winequality-white.csv

red = np.loadtxt("winequality-red.csv", delimiter=";", skiprows=1)
white = np.loadtxt("winequality-white.csv", delimiter=";", skiprows=1)
pooled = np.vstack([red, white])
colour = np.concatenate([np.ones(len(red)), np.zeros(len(white))])
wine_features = np.column_stack([pooled[:, :-1], colour])
wine_targets = pooled[:, -1]

# The model has no bias term, so the features are standardised and the target centred, both with
# the training statistics only. A fifth of the objects is held out for validation.
order = np.random.default_rng(42).permutation(len(wine_features))
train = order[: round(0.8 * len(order))]
validation = order[len(train) :]
features = (wine_features - wine_features[train].mean(axis=0)) / wine_features[train].std(axis=0)
centred_targets = wine_targets - wine_targets[train].mean()
X_train, y_train = features[train], centred_targets[train]
X_val, y_val = features[validation], centred_targets[validation]
print(X_train.shape, X_val.shape)
```

```python
import time


def matrix_formula(X, y, reg_coeff=0.0):
    gram = X.T.dot(X) + reg_coeff * len(X) * np.eye(X.shape[1])
    return np.linalg.inv(gram).dot(X.T.dot(y))


def gd_step_size(X, reg_coeff=0.0):
    # 1 / largest eigenvalue of the Hessian (2 / n) X^T X + 2 * reg_coeff * I, which is the largest
    # step full-batch descent converges at. An L1 term adds no curvature.
    return 1.0 / (2.0 * np.linalg.eigvalsh(X.T.dot(X) / len(X))[-1] + 2.0 * reg_coeff)


def fit_gd(X, y, loss_derivative, reg_derivative=None, reg_coeff=0.0, step=0.1, n_steps=1000):
    # The checker's own loop: no gradient clipping and no logging, so the numbers match its report.
    w = np.zeros(X.shape[1])
    for _ in range(n_steps):
        gradient = loss_derivative(X, y, w)
        if reg_derivative is not None:
            gradient = gradient + reg_coeff * reg_derivative(w)
        w = w - step * gradient
    return w


def fit_sgd(X, y, loss_derivative, step=0.1, decay=400.0, epochs=20, batch_size=64, seed=0):
    rng = np.random.default_rng(seed)
    w = np.zeros(X.shape[1])
    taken = 0
    for _ in range(epochs):
        order = rng.permutation(len(X))
        for start in range(0, len(order), batch_size):
            batch = order[start : start + batch_size]
            w = w - step / (1.0 + taken / decay) * loss_derivative(X[batch], y[batch], w)
            taken += 1
    return w
```

```python
REG_COEFF = 0.02
step = gd_step_size(X_train)
reg_step = gd_step_size(X_train, REG_COEFF)
losses = LossAndDerivatives


def timed(fit):
    started = time.perf_counter()
    weights = fit()
    return weights, time.perf_counter() - started


solvers = [
    ("matrix formula, np.linalg.inv", lambda: matrix_formula(X_train, y_train)),
    ("matrix formula, ridge", lambda: matrix_formula(X_train, y_train, REG_COEFF)),
    ("gradient descent, MSE", lambda: fit_gd(X_train, y_train, losses.mse_derivative, step=step)),
    (
        "gradient descent, MSE + L2",
        lambda: fit_gd(
            X_train, y_train, losses.mse_derivative, losses.l2_reg_derivative, REG_COEFF, reg_step
        ),
    ),
    (
        "gradient descent, MSE + L1",
        lambda: fit_gd(
            X_train, y_train, losses.mse_derivative, losses.l1_reg_derivative, REG_COEFF, step
        ),
    ),
    (
        "gradient descent, MAE",
        lambda: fit_gd(X_train, y_train, losses.mae_derivative, step=0.1, n_steps=2000),
    ),
    ("stochastic gradient descent, MSE", lambda: fit_sgd(X_train, y_train, losses.mse_derivative)),
]

results = [(label, *timed(solver)) for label, solver in solvers]
w_reference = results[0][1]

print(f"{'method':<34}{'time, ms':>10}{'|w - w*| / |w*|':>18}{'val MSE':>10}{'val MAE':>10}")
for label, w, seconds in results:
    residual = X_val.dot(w) - y_val
    relative = np.linalg.norm(w - w_reference) / np.linalg.norm(w_reference)
    print(
        f"{label:<34}{seconds * 1e3:>10.2f}{relative:>18.2e}"
        f"{np.mean(residual**2):>10.4f}{np.mean(np.abs(residual)):>10.4f}"
    )
```

Three things are worth reading off the table.

The matrix formula is both the fastest and the exact answer here, because $d = 12$ makes the
$O(nd^2 + d^3)$ inverse trivial; it is gradient descent that becomes the only option as $d$ grows.
Descent is not a cheaper way to the same place — it is the only way once the place cannot be
written down.

SGD reaches a neighbourhood of $w^*$ in a fraction of the full-batch time, and its validation MSE
is slightly *better* than the exact solution's. That is not a mistake: the noise in the stochastic
step keeps it from committing to the training set, which is implicit regularisation.

Each run converges to the minimiser of its own objective. The L2 run matches the ridge formula
rather than $w^*$; the L1 run drives the weakest weights onto zero, which L2 never does; and the
MAE run loses on validation MSE and beats least squares on validation MAE. Try changing
`REG_COEFF`, `n_steps` and `batch_size` and watch which column moves.
